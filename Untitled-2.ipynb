```python
%pip install scikit-learn

# Import necessary libraries and configure the environment
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')  # Use Agg backend as required
import matplotlib.pyplot as plt
plt.switch_backend('Agg')  # Ensuring non-interactive backend when only plt is imported
get_ipython().run_line_magic('matplotlib', 'inline')

import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

# For reproducibility
RANDOM_STATE = 42

sns.set(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Load the dataset
data_path = 'C:\\Users\\annar\\OneDrive\\Escritorio\\Popular_Spotify_Songs (1).csv'
df = pd.read_csv(data_path, encoding='ISO-8859-1')

# Quick look at the data structure
print('Dataset shape:', df.shape)
print(df.head())
```

```python
# Display the initial data types
print('Initial Data Types:')
print(df.dtypes)

# Clean the 'streams' column: remove commas and convert to numeric, coercing errors to NaN
df['streams'] = pd.to_numeric(df['streams'].str.replace(',', '', regex=False), errors='coerce')

# Convert columns that represent numbers but are stored as strings
df['in_deezer_playlists'] = pd.to_numeric(df['in_deezer_playlists'], errors='coerce')
df['in_shazam_charts'] = pd.to_numeric(df['in_shazam_charts'], errors='coerce')

# Fill missing numeric values with 0
df.fillna(0, inplace=True)

# Combine released_year, released_month, and released_day into a 'release_date' column
# Using a dictionary with keys 'year', 'month', 'day' to meet requirements of pd.to_datetime
df['release_date'] = pd.to_datetime(
    dict(year=df['released_year'], month=df['released_month'], day=df['released_day']),
    errors='coerce'
)

# Optional: Drop individual date columns if they are no longer needed
# df.drop(['released_year', 'released_month', 'released_day'], axis=1, inplace=True)

print('\nData Types after cleaning:')
print(df.dtypes)
```

```python
# Display summary statistics for numeric columns
print('Summary Statistics:')
print(df.describe())

# Histogram for the 'streams' column to inspect its distribution
sns.histplot(df['streams'], kde=True, color='blue')
plt.title('Distribution of Streams')
plt.xlabel('Streams')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# Count plot for the 'mode' column to see the distribution of musical modes
sns.countplot(data=df, x='mode', palette='pastel')
plt.title('Count of Modes')
plt.xlabel('Mode')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Select only numeric columns
numeric_df = df.select_dtypes(include=[np.number])

if numeric_df.shape[1] >= 4:
    corr = numeric_df.corr()
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Correlation Heatmap of Numeric Features')
    plt.tight_layout()
    plt.show()
else:
    print('Not enough numeric features for a correlation heatmap.')

# Pairplot to visualize pairwise relationships among a subset of numeric features
selected_features = ['streams', 'in_spotify_playlists', 'in_apple_playlists', 'bpm']
sns.pairplot(df[selected_features])
plt.suptitle('Pairplot of Selected Features', y=1.02)
plt.show()
```
